# DateTime

`pd.date_range`function get value in format **YYYY-MM-DD** and in **string format**

```python
df2 = pd.date_range('20220129', periods=6)

df2
```

**OR**

```python
df4 = pd.date_range(start='20220129', end='20220203')

df4
```

```python
# wrong declaration
df3 = pd.date_range(20220129, periods=6)

df3
```

```python
df5 = pd.DataFrame(np.random.randn(6,4), index=df2, columns= list('ABCD'))

df5
```

Creating dataframe of random numbers in 6 by 4 matrix, index is given as df2 which contains 6 dates, column name is given as A, B, C, D

```python
df6 = pd.DataFrame({'A': 1,
                   'B': pd.Timestamp('20220101'),
                   'C': pd.Series(1, index=list(range(4)), dtype=float),
                   'D': np.array([3] * len(df6), dtype=int ),
                   'E': pd.Categorical(['Test', 'Train', 'View', 'Final']),
                   'F': 'Foo'})

df6
```
