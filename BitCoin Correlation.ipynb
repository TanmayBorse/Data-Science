```python
import pandas as pd
import pytrends
from pytrends.request import TrendReq
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
pytrends = TrendReq(hl= 'en-US', tz = 120)
```

```python
kw_words = ['bitcoin', 'crypto']
pytrends.build_payload(kw_words)
```

```python
data_request = pytrends.interest_over_time()
data_request.index = data_request.index.date
```

```python
data_request.head()
```

```python
data_request.shape
```

```python
data_request.iloc[:, : -1].plot(x = None, y= None, figsize=(12,9))
plt.title('Rrequest rate worldwide', fontsize = 16)
plt.ylabel('Coefficient', fontsize=14)
plt.ylabel('Date', fontsize=14)
plt.grid(which='major', color = 'k', linestyle = '-', linewidth = 0.5)
plt.show();
```
